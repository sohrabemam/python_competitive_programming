```python
#palindrome
def longest_palindrome(s):
    longest= ""
    for i in range(len(s)):
        for j in range(i, len(s)):
            substring = s[i:j+1]
            if substring == substring[::-1] and len(substring) > len(longest):
                longest = substring
    return longest
```

```python
a = "j"
print(longest_palindrome(a))
```

```python
def rec_fib(n, dp):
    if n ==0 or n ==1:
        return n
    if dp[n] != -1:
        return dp[n]
    dp[n] = rec_fib(n-1, dp) + rec_fib(n-2, dp)
    return dp[n]
```

```python
n = int(input("Enter a number: "))
dp = [-1]* (n+1)
print(dp)
print(rec_fib(n, dp))
```
